```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
from scipy.ndimage import zoom # for rescaling
from scipy.ndimage import convolve # for blurring/smoothing
import math

image_name = input("Enter image name: ")
img = Image.open("images/" + image_name).convert("L")
img_array = np.array(img)
```

```python
# just show regular image
plt.imshow(img_array, cmap='gray')
plt.axis('on')
plt.show()
```

```python
# rescale image

# set scaling factor
scaling_factor = 0.5

scaled_img = zoom(img_array, (scaling_factor, scaling_factor))

# display regular image
plt.imshow(img_array, cmap='gray')
plt.axis('on')
plt.show()

# display rescaled image
plt.imshow(scaled_img, cmap='gray')
plt.axis('on')
plt.show()
```

```python
# rotate image

# select whatever rotation wanted
rotated_img = img.rotate(180) 

rotated_img_array = np.array(rotated_img)

# display rotated image
plt.imshow(rotated_img_array, cmap='gray')
plt.axis('off')
plt.show()
```

```python
# blurring:

# box blur
kernel = np.ones((3, 3))/9

# gaussian blur
kernel2 = np.array([
    [ 1/16, 2/16,  1/16],
    [2/16,  4/16, 2/16],
    [ 1/16, 2/16,  1/16]
])

# apply convolution
blurred_img_array = convolve(img_array, kernel)
blurred_img_array2 = convolve(img_array, kernel2)


# display regular image
plt.imshow(img_array, cmap='gray')
plt.axis('off')
plt.show()

# display blurred image
plt.imshow(blurred_img_array, cmap='gray')
plt.axis('off')
plt.show()

# plt.imshow(blurred_img_array2, cmap='gray')
# plt.axis('off')
# plt.show()
```

```python
# sharpen:

# standard sharpening kernel
kernel = np.array([
    [ 0, -1,  0],
    [-1,  5, -1],
    [ 0, -1,  0]
])

# apply convolution
blurred_img_array = convolve(img_array, kernel)

# display regular image
plt.imshow(img_array, cmap='gray')
plt.axis('off')
plt.show()

# display blurred image
plt.imshow(blurred_img_array, cmap='gray')
plt.axis('off')
plt.show()
```

```python
# horizontal edge detection:

kernel = np.array([
    [-1, -2, -1],
    [ 0,  0,  0],
    [ 1,  2,  1]
])

# apply convolution
blurred_img_array = convolve(img_array, kernel)

# display regular image
plt.imshow(img_array, cmap='gray')
plt.axis('off')
plt.show()

# display blurred image
plt.imshow(blurred_img_array, cmap='gray')
plt.axis('off')
plt.show()
```

```python
# vertical edge detection:

kernel = np.array([
    [-1, 0, 1],
    [-2, 0, 2],
    [-1, 0, 1]
])

# apply convolution
blurred_img_array = convolve(img_array, kernel)

# display regular image
plt.imshow(img_array, cmap='gray')
plt.axis('off')
plt.show()

# display blurred image
plt.imshow(blurred_img_array, cmap='gray')
plt.axis('off')
plt.show()
```
